# Logistic Regression 

## Step1 - Importing all the required libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix,ConfusionMatrixDisplay

import pickle
```

## Step2 - Loading the dataset using pandas 

```python
data=pd.read_csv(r"C:\Users\HP\Downloads\archive (15)\diabetes_prediction_dataset.csv")
```

## Step4 - Selecting which features must be taken as the dependent and the independent variable

```python
X=data[['smoking_history','age','blood_glucose_level','HbA1c_level']]
Y=data['diabetes']
```

```python
print(X)
print(Y)
```

## Step6 - Using one hot encoding in order to convert the categorical variables into numerical 

```python
X=pd.get_dummies(X, columns=['smoking_history'],drop_first=True)
```

## Step 7 - Splitting the data into testing and training data

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

## Step8 - Importing Logistic Regression

```python
classifier=LogisticRegression(class_weight='balanced')
```

## Step9 - Fitting the data in order to train the model

```python
classifier.fit(X_train,Y_train)
```

## Step10 - Predicting the values 

```python
prediction=classifier.predict(X_test)
```

## Step11 - Evaluation of the model

```python
from sklearn.metrics import (accuracy_score,precision_score,recall_score,f1_score)
```

```python
print(f"\033[1mF1 Score:\033[0m", f1_score(Y_test, prediction))
print(f"\033[1mAccuracy:\033[0m", accuracy_score(Y_test, prediction))
print(f"\033[1mRecall:\033[0m",recall_score(Y_test,prediction))
print(f"\033[1mPrecision:\033[0m", precision_score(Y_test, prediction))
```

## Step12 - Converting this into the pkl file

```python
with open("diabetes.pkl", "wb") as file:
    pickle.dump(classifier, file)
print("Model saved successfully")
```
