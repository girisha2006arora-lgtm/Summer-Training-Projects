```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error,root_mean_squared_error

import pickle
```

```python
df=pd.read_csv(r"C:\Users\HP\Downloads\Restaurant_revenue (1).csv")
```

```python
## data cleaning
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.duplicated()
```

```python
X = df[['Number_of_Customers',
        'Menu_Price',
        'Marketing_Spend',
        'Average_Customer_Spending',
        'Promotions',
        'Reviews']]

Y = df['Monthly_Revenue']
```

```python
print(X)
print(Y)
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
reg=RandomForestRegressor(n_estimators=100,random_state=42)
```

```python
reg.fit(X_train,Y_train)
```

```python
Y_pred=reg.predict(X_test)
```

```python
print("R2 Score: ",r2_score(Y_test,Y_pred))
print("Mean square error: ",mean_squared_error(Y_test,Y_pred))
print("Mean absolute error: ",mean_absolute_error(Y_test,Y_pred))
print("root_mean_squared_error: ",root_mean_squared_error(Y_test,Y_pred))
```

```python
from sklearn import tree
plt.figure(figsize=(30,20))
tree.plot_tree(
    reg.estimators_[0],
    filled=True,
    feature_names=X.columns,
    rounded=True
)
plt.show()
```

```python
with open("Restaurant_revenue.pkl", "wb") as file:
    pickle.dump(reg, file)
print("Model saved successfully")
```

```python
from tkinter import *
import pickle
import numpy as np

# ------------------ Load Model ------------------
model = pickle.load(open("Restaurant_revenue.pkl", "rb"))

root = Tk()
root.title("Restaurant Revenue Prediction System")
root.geometry("850x650")
root.configure(bg="#4B0082")      


heading = Label(
    root,
    text="🍽 Restaurant Revenue Prediction System",
    font=("Georgia",22,"bold"),
    bg="#4B0082",
    fg="white"
)
heading.pack(pady=20)


frame = Frame(root,bg="#E6E6FA",bd=5,relief=RIDGE)
frame.pack(pady=10)

labels = [
    "Number of Customers",
    "Menu Price",
    "Marketing Spend",
    "Average Customer Spending",
    "Promotions (0/1)",
    "Reviews"
]

entries = []

for i,text in enumerate(labels):

    Label(
        frame,
        text=text,
        font=("Arial",13,"bold"),
        bg="#E6E6FA",
        fg="#4B0082"
    ).grid(row=i,column=0,padx=20,pady=12,sticky="w")

    e = Entry(
        frame,
        width=25,
        font=("Arial",13),
        bd=3,
        relief=GROOVE
    )
    e.grid(row=i,column=1,padx=20,pady=12)

    entries.append(e)

def predict():
    num_customers = float(entries[0].get())
    menu_price = float(entries[1].get())
    marketing_spend = float(entries[2].get())
    avg_customer_spending = float(entries[3].get())
    promotions = int(entries[4].get())
    reviews = int(entries[5].get())

    prediction = reg.predict([[
        num_customers,
        menu_price,
        marketing_spend,
        avg_customer_spending,
        promotions,
        reviews
    ]])

    output.config(text=f"Predicted Monthly Revenue : ₹ {prediction[0]:,.2f}")
Button(
    root,
    text="Predict Revenue",
    font=("Arial",15,"bold"),
    bg="#800080",
    fg="white",
    padx=20,
    pady=8,
    command=predict
).pack(pady=20)

output = Label(
    root,
    text="Predicted Monthly Revenue",
    font=("Arial",18,"bold"),
    bg="#4B0082",
    fg="#FFD700"
)
output.pack(pady=15)

footer = Label(
    root,
    text="Developed using Random Forest Regressor",
    font=("Arial",11),
    bg="#4B0082",
    fg="#D8BFD8"
)
footer.pack(side=BOTTOM,pady=10)

root.mainloop()
```
